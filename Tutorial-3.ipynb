## Setup
------

```python
# Import necessary libraries
import numpy as np
import mbuild as mb
from mbuild.polymer import Polymer
from mbuild.path.build import Path, lamellar

# Check mBuild version
print(f"mBuild version: {mb.__version__}")
```

## Creating a path
-----


```python
# Create a path
path = Path()

# Add points to path
points = np.random.rand(10,3)
path.append_coordinates(points)

# Visualize path
path.visualize(radius=0.15)
```

```python
# Build up a more complicated path with edges
path = Path()

import numpy as np

num_points = [5, 3, 2]
names = ["_A", "_B", "_C"]
step = 2
x_val = 0

for name, npoint in zip(names, num_points):
    x = np.arange(1, 1 + npoint) * step
    y = np.full(npoint, x_val) * step
    z = np.zeros(npoint)
    
    points = np.stack([x, y, z], axis=-1)
    path.append_coordinates(points, name)
    x_val += 1

# connect all points with the same name together in a line
path._connect_edges("linear", np.arange(5))
path._connect_edges("linear", np.arange(5,8))
path._connect_edges("linear", np.arange(8,10))

path.visualize(radius=1)
```

## Path templates
--------

```python
# Path subclasses
from mbuild.path.build import Path, lamellar, straight_line, cyclic, knot, helix, zigzag, spiral_2D

# Generate Paths
path = Path()
lamellar(path, layer_separation=0.3, layer_length=3, num_layers=5, bond_length=0.15, bead_name="_A")
n = len(path.coordinates)
straight_line(path, spacing=0.15, N=20, direction=(0,1,0), bead_name="_B")
m = len(path.coordinates)
path.coordinates[n:][:,0] += 3
cyclic(path, spacing=0.14, N=30, bead_name="_C")
l = len(path.coordinates)
path.coordinates[m:][:,0] += 5
path.coordinates[m:][:,1] += 1

# Visualize
path.visualize(radius=0.1)
```

## Create a lamellar polymer
-----

```python
# Load monomer
from mbuild.path.build import lamellar

# Create Path
path = lamellar(num_layers=5, layer_separation=0.35, layer_length=6, bond_length=0.2)
# path.visualize(radius=.1) # visualize just the path

# Polymer from path
polymer = Polymer()
polymer.add_monomer(
    mb.load("C{head}C{tail}", smiles=True),
    head_tag="head",
    tail_tag="tail",
    separation=0.2
)
polymer.build_from_path(path)
polymer.visualize()
```
